# 위험유형별 Top10

```python
# 위험유형별 Top 10 분석
# 위험유형별로 가장 위험도가 높은 역, 호선, 시간대를 확인
def get_top10_by_risk_type(risk_type):
    result = (
        df[df['위험유형'] == risk_type]
        .sort_values('Risk_Score', ascending=False)
        .head(10)
        .reset_index(drop=True)
    )

    display_cols = [
        '역명_clean',
        '호선명',
        'time_group',
        'Risk_Score',
        '위험유형',
        '주요위험원인',
        '추천대응전략'
    ]

    result_table = result[display_cols].copy()
    result_table.insert(0, '순위', range(1, len(result_table) + 1))

    result_table['주요위험원인'] = (
        result_table['주요위험원인']
        .astype(str)
        .str.replace('형', ' 요인')
    )

    result_table['Risk_Score'] = result_table['Risk_Score'].round(1)

    return result_table
```

# 시간대별 위험 패턴

```python
# 시간대별 위험 패턴 분석
# 시간대별 평균 Risk Score와 고위험 구간 수를 확인
time_pattern = (
    df.groupby('time_group')
      .agg(
          평균위험도=('Risk_Score', 'mean'),
          최대위험도=('Risk_Score', 'max'),
          구간수=('Risk_Score', 'count'),
          고위험구간수=('Risk_Score', lambda x: (x >= 80).sum())
      )
      .reset_index()
)

time_pattern['평균위험도'] = time_pattern['평균위험도'].round(1)
time_pattern['최대위험도'] = time_pattern['최대위험도'].round(1)

time_pattern
```

```python
# 해석:
# 시간대별 평균위험도는 낮시간이 25.5점으로 가장 높게 나타났고,
# 퇴근시간 24.2점, 출근시간 23.0점, 기타 17.1점 순으로 나타났다.
#
# 고위험구간수는 낮시간이 29개로 가장 많았고,
# 퇴근시간이 21개, 출근시간이 10개, 기타가 6개로 나타났다.
#
# 이는 위험이 특정 출퇴근 시간에만 집중되는 것이 아니라,
# 낮시간에도 환승·구조·유입 요인이 복합적으로 작용해
# 고위험 구간이 많이 나타난다는 것을 의미한다.
#
# 따라서 안전대책은 출퇴근 시간 중심의 일시적 대응뿐 아니라,
# 낮시간 주요 환승역과 구조위험역까지 포함하는 상시 관리 전략이 필요하다.
```

# What-if 정책 민감도 분석
## 가정 기반 시뮬레이션
- 목적: 특정 위험요인이 완화된다고 가정했을 때, Risk Score가 얼마나 낮아지는지 확인

```python
# What-if 정책 민감도 분석 함수
# 특정 위험요인이 일정 비율 완화된다고 가정했을 때
# Risk Score 변화량을 계산

def what_if_analysis(station, line, time_group, factor, reduction_rate=0.1):
    result = df[
        (df['역명_clean'] == station) &
        (df['호선명'] == line) &
        (df['time_group'] == time_group)
    ]

    if result.empty:
        print("해당 조건의 결과가 없습니다.")
        return

    row = result.iloc[0]

    factor_col_map = {
        '환승': '환승기여도',
        '구조': '구조기여도',
        '혼잡': '혼잡기여도',
        '유입': '유입기여도'
    }

    if factor not in factor_col_map:
        print("factor는 '환승', '구조', '혼잡', '유입' 중 하나로 입력하세요.")
        return

    current_score = row['Risk_Score']
    contribution = row[factor_col_map[factor]]

    # 가정 기반 계산:
    # 해당 요인 기여도만큼 Risk Score에 영향을 준다고 보고,
    # 그 요인을 reduction_rate만큼 완화했을 때의 점수 감소분 계산
    score_reduction = current_score * (contribution / 100) * reduction_rate
    new_score = current_score - score_reduction

    print("=" * 70)
    print("🚇 MetroGuard-AI What-if 정책 민감도 분석")
    print("=" * 70)
    print(f"분석 대상: {station} / {line} / {time_group}")
    print(f"현재 Risk Score: {current_score:.1f}점")
    print()
    print(f"완화 가정 요인: {factor} 요인")
    print(f"해당 요인 기여도: {contribution:.1f}%")
    print(f"완화 가정 비율: {reduction_rate * 100:.0f}%")
    print()
    print(f"예상 Risk Score 감소폭: {score_reduction:.1f}점")
    print(f"What-if 적용 후 Risk Score: {new_score:.1f}점")
    print()
    print("해석:")
    print("본 결과는 실제 정책 효과를 단정하는 것이 아니라,")
    print("특정 위험요인이 완화된다는 가정하에 Risk Score 변화 방향을 확인하는 민감도 분석입니다.")
    print("=" * 70)
```

```python
what_if_analysis('사당', '2호선', '퇴근시간', '환승', reduction_rate=0.1)
```

```python
# 최종: 드롭다운 기반 What-if 정책 민감도 분석 시스템

import ipywidgets as widgets
from IPython.display import display, clear_output, Markdown

# 시스템 설명
display(Markdown("""
# 🧪 MetroGuard-AI What-if 정책 민감도 분석

이 시스템은 실제 정책 효과를 단정하지 않고,
특정 위험요인이 일정 비율 완화된다고 가정했을 때
**Risk Score가 얼마나 변하는지 확인하는 가정 기반 민감도 분석**입니다.

- **완화요인**: 정책을 통해 줄인다고 가정할 위험요인
  - 환승: 환승 동선 복잡도 완화
  - 구조: 승강장 구조·시설 위험 완화
  - 혼잡: 승하차 인원 및 승강장 밀집 완화
  - 유입: 역 주변 외부 유입 인구 집중 완화

- **완화비율**
  - 10% 완화: 낮은 강도 개선 시나리오
  - 20% 완화: 중간 강도 개선 시나리오
  - 30% 완화: 높은 강도 개선 시나리오

※ 실제 정책 효과가 아니라, 위험요인 변화에 따른 Risk Score 민감도를 보기 위한 분석입니다.
"""))

# 선택지 설정
station_options = sorted(df['역명_clean'].dropna().unique())

factor_options = [
    ('환승 요인 완화', '환승'),
    ('구조 요인 완화', '구조'),
    ('혼잡 요인 완화', '혼잡'),
    ('유입 요인 완화', '유입')
]

reduction_options = {
    '10% 완화 / 낮은 강도 개선': 0.1,
    '20% 완화 / 중간 강도 개선': 0.2,
    '30% 완화 / 높은 강도 개선': 0.3
}

# 드롭다운 UI
station_dropdown_w = widgets.Dropdown(
    options=station_options,
    description='역명:',
    layout=widgets.Layout(width='360px')
)

line_dropdown_w = widgets.Dropdown(
    options=[],
    description='호선:',
    layout=widgets.Layout(width='360px')
)

time_dropdown_w = widgets.Dropdown(
    options=[],
    description='시간대:',
    layout=widgets.Layout(width='360px')
)

factor_dropdown = widgets.Dropdown(
    options=factor_options,
    description='완화요인:',
    layout=widgets.Layout(width='360px')
)

reduction_dropdown = widgets.Dropdown(
    options=reduction_options,
    description='완화비율:',
    layout=widgets.Layout(width='360px')
)

button_w = widgets.Button(
    description='What-if 분석',
    button_style='warning',
    icon='flask',
    layout=widgets.Layout(width='200px')
)

output_w = widgets.Output()


# 역명 선택 시 해당 역의 호선만 표시
def update_line_options_w(*args):
    station = station_dropdown_w.value

    lines = sorted(
        df[df['역명_clean'] == station]['호선명']
        .dropna()
        .unique()
    )

    line_dropdown_w.options = lines

    if len(lines) > 0:
        line_dropdown_w.value = lines[0]

    update_time_options_w()


# 역명 + 호선 선택 시 해당 조건의 시간대만 표시
def update_time_options_w(*args):
    station = station_dropdown_w.value
    line = line_dropdown_w.value

    times = sorted(
        df[
            (df['역명_clean'] == station) &
            (df['호선명'] == line)
        ]['time_group']
        .dropna()
        .unique()
    )

    time_dropdown_w.options = times

    if len(times) > 0:
        time_dropdown_w.value = times[0]


# 버튼 클릭 시 What-if 분석 실행
def on_button_click_w(b):
    with output_w:
        clear_output()

        station = station_dropdown_w.value
        line = line_dropdown_w.value
        time_group = time_dropdown_w.value
        factor = factor_dropdown.value
        reduction_rate = reduction_dropdown.value

        what_if_analysis(
            station,
            line,
            time_group,
            factor,
            reduction_rate=reduction_rate
        )


# 이벤트 연결
station_dropdown_w.observe(update_line_options_w, names='value')
line_dropdown_w.observe(update_time_options_w, names='value')
button_w.on_click(on_button_click_w)

# 초기값 설정
update_line_options_w()

# UI 출력
display(
    station_dropdown_w,
    line_dropdown_w,
    time_dropdown_w,
    factor_dropdown,
    reduction_dropdown,
    button_w,
    output_w
)
```
