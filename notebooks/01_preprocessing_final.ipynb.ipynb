````python
df = 승하차인원.copy()

# 혹시 공백 제거
df.columns = df.columns.str.strip()

# 시간대 컬럼 (승차인원, 하차인원)
time_cols = [col for col in df.columns if ('승차인원' in col) or ('하차인원' in col)]

print(len(time_cols))
print(time_cols[:5])
````

````python
long_df = df.melt(
    id_vars=['사용월', '호선명', '지하철역'],
    value_vars=time_cols,
    var_name='time_type',
    value_name='count'
)
````

````python
# 컬럼이 숫자가 아니라 문자열처럼 이어붙은 값을 깨진거 처리하기
long_df['count'] = (
    long_df['count']
    .astype(str)
    .str.replace(',', '', regex=False)
)

long_df['count'] = pd.to_numeric(long_df['count'], errors='coerce').fillna(0)
````

````python
long_df['hour'] = long_df['time_type'].str.extract(r'(\d{2})시').astype(int)
````

````python
def make_time_group(hour):
    if 7 <= hour < 9:
        return '출근시간'
    elif 9 <= hour < 16:
        return '낮시간'
    elif 16 <= hour < 20 :
        return '퇴근시간'
    else:
        return '기타'

long_df['time_group'] = long_df['hour'].apply(make_time_group)
````

````python
승하차인원_전처리 = (
    long_df
    .groupby(['사용월', '호선명', '지하철역', 'time_group'], as_index=False)
    .agg(혼잡=('count', 'sum'))
)
````

사고 데이터셋 전처리 : 역별, 시간대별로 사고 수 집계함. 다른 데이터셋과의 merge위해 역명과 호선 통일함

````python
사고['hour'] = 사고['발생시간'].str.split(':').str[0].astype(int)

def make_time_group(hour):
    if 7 <= hour < 9:
        return '출근시간'
    elif 9 <= hour < 16:
        return '낮시간'
    elif 16 <= hour < 20:
        return '퇴근시간'
    else:
        return '기타'

사고['time_group'] = 사고['hour'].apply(make_time_group)
````

````python
# 날짜 처리
사고['발생일'] = pd.to_datetime(사고['발생일'], errors='coerce')

사고['사용월'] = 사고['발생일'].dt.strftime('%Y%m').astype(int)

# 역명 정리 함수 정의
import re

def clean_station_name(x):
    if pd.isna(x):
        return x

    x = str(x)
    x = re.sub(r'\(\d+\)', '', x)  # 교대(2)역 -> 교대역
    x = x.replace('역', '')        # 교대역 -> 교대
    x = x.strip()

    return x
````

````python
import numpy as np
사고['역명_clean'] = 사고['발생역'].apply(clean_station_name)

# 호선 추출 (예: 교대(2)역 → 2호선)
사고['호선명'] = 사고['발생역'].str.extract(r'\((\d+)\)').astype(float).astype('Int64').astype(str) + '호선'

# 없는 경우 처리
사고['호선명'] = 사고['호선명'].replace('<NA>호선', np.nan)
````

````python
사고= (
    사고
    .groupby(['사용월', '역명_clean', '호선명', 'time_group'], as_index=False)
    .size()
    .rename(columns={'size': '사고수'})
)
````

````python
# 역명 통일 부족한 것 역명 통일시키기
import re

def clean_station_name(x):
    x = str(x).strip()
    x = x.replace(' ', '')
    x = x.replace('（', '(').replace('）', ')')
    x = re.sub(r'\(.*?\)', '', x)
    x = x.replace('역', '')
    return x

승하차인원['역명_clean'] = 승하차인원['지하철역'].apply(clean_station_name)
이격거리['역명_clean'] = 이격거리['역명'].apply(clean_station_name)
환승역['역명_clean'] = 환승역['역명'].apply(clean_station_name)
사고['역명_clean'] = 사고['역명_clean'].apply(clean_station_name)

# 동대문역사문화공원 -> 동대문사문화공원 으로 생성되는 오류 발생
````

````python
import re

def clean_station_name(x):
    x = str(x).strip()
    x = x.replace(' ', '')
    x = x.replace('（', '(').replace('）', ')')

    # 괄호 안 내용 제거: 교대(2)역 -> 교대역
    x = re.sub(r'\(.*?\)', '', x)

    # 문자열 맨 끝의 '역'만 제거
    x = re.sub(r'역$', '', x)

    return x
````
